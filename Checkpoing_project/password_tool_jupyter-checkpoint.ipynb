# 🔐 비밀번호 생성기 & 강도 체커
표준 라이브러리(`random`, `string`, `re`)만 사용합니다.

```python
import random
import string
import re

UPPER   = string.ascii_uppercase
LOWER   = string.ascii_lowercase
DIGITS  = string.digits
SYMBOLS = '!@#$%^&*()_+-=[]{}|;:,.<>?'

def generate_password(length=16, use_upper=True, use_lower=True,
                      use_digits=True, use_symbols=True):
    if length < 4:
        raise ValueError('비밀번호 길이는 최소 4자 이상이어야 합니다.')
    pool, required = '', []
    if use_upper:   pool += UPPER;   required.append(random.choice(UPPER))
    if use_lower:   pool += LOWER;   required.append(random.choice(LOWER))
    if use_digits:  pool += DIGITS;  required.append(random.choice(DIGITS))
    if use_symbols: pool += SYMBOLS; required.append(random.choice(SYMBOLS))
    if not pool:
        raise ValueError('최소 하나의 문자 유형을 선택해야 합니다.')
    rest = [random.choice(pool) for _ in range(length - len(required))]
    pw = required + rest
    random.shuffle(pw)
    return ''.join(pw)

def check_strength(password):
    details = {
        '8자 이상'     : len(password) >= 8,
        '12자 이상'    : len(password) >= 12,
        '대문자 포함'  : bool(re.search(r'[A-Z]', password)),
        '소문자 포함'  : bool(re.search(r'[a-z]', password)),
        '숫자 포함'    : bool(re.search(r'\d', password)),
        '특수문자 포함': bool(re.search(r'[!@#$%^&*()\-_=+\[\]{}|;:,.<>?]', password)),
    }
    score = sum(details.values())
    grade = ['매우 약함','매우 약함','매우 약함','약함','보통','강함','매우 강함'][score]
    return {'score': score, 'grade': grade, 'details': details}

print('✅ 함수 로드 완료')
```

## 1. 비밀번호 생성
아래 변수를 바꾸고 셀을 실행하세요.

```python
# ── 여기를 수정하세요 ──────────────────────
LENGTH      = 16
USE_UPPER   = True
USE_LOWER   = True
USE_DIGITS  = True
USE_SYMBOLS = True
# ──────────────────────────────────────────

pw = generate_password(LENGTH, USE_UPPER, USE_LOWER, USE_DIGITS, USE_SYMBOLS)
result = check_strength(pw)

bar = '█' * result['score'] + '░' * (6 - result['score'])
print(f'생성된 비밀번호 : {pw}')
print(f'강도            : [{bar}] {result["grade"]} ({result["score"]}/6)')
print()
for cond, passed in result['details'].items():
    print(f'  {"✔" if passed else "✘"}  {cond}')
```

## 2. 내 비밀번호 강도 체크

```python
# ── 확인할 비밀번호를 여기에 입력하세요 ──
MY_PASSWORD = 'MyPass123!'
# ─────────────────────────────────────────

result = check_strength(MY_PASSWORD)
bar = '█' * result['score'] + '░' * (6 - result['score'])
print(f'강도 : [{bar}] {result["grade"]} ({result["score"]}/6)')
print()
for cond, passed in result['details'].items():
    print(f'  {"✔" if passed else "✘"}  {cond}')
```

## 3. 한 번에 여러 개 생성

```python
COUNT = 5  # 생성할 개수

print(f'길이 16 비밀번호 {COUNT}개')
print('-' * 35)
for i in range(COUNT):
    pw = generate_password(16)
    grade = check_strength(pw)['grade']
    print(f'{i+1:2}.  {pw}  ({grade})')
```
