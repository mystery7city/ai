## 6 Data Cleaning

```python
def make_long(df, value_name):
    df_long = df.melt(
        id_vars=["Country Name", "Country Code", "Indicator Name", "Indicator Code"],
        var_name="year",
        value_name=value_name
    )
    
    df_long = df_long[["Country Name", "Country Code", "year", value_name]]
    df_long = df_long.rename(columns={
        "Country Name": "country",
        "Country Code": "code"
    })
    
    df_long["year"] = pd.to_numeric(df_long["year"], errors="coerce")
    df_long = df_long.dropna(subset=["year"])
    df_long["year"] = df_long["year"].astype(int)
    
    return df_long
```

### 각 파일을 세로형으로

```python
gdp_long = make_long(gdp, "gdp")
exports_long = make_long(exports, "exports_usd")
hightech_long = make_long(hightech, "hightech_exports")
military_usd_long = make_long(military_usd, "military_usd")
military_per_gdp_long = make_long(military_per_gdp, "military_per_gdp")
patent_long = make_long(patent, "patent_residents")
population_long = make_long(population, "population")
rd_long = make_long(rd, "rd_spending")
```

### 나라와 연도 기준으로 합치기

```python
df = gdp_long.merge(exports_long, on=["country", "code", "year"], how="outer")
df = df.merge(hightech_long, on=["country", "code", "year"], how="outer")
df = df.merge(military_usd_long, on=["country", "code", "year"], how="outer")
df = df.merge(military_per_gdp_long, on=["country", "code", "year"], how="outer")
df = df.merge(patent_long, on=["country", "code", "year"], how="outer")
df = df.merge(population_long, on=["country", "code", "year"], how="outer")
df = df.merge(rd_long, on=["country", "code", "year"], how="outer")
```

```python
import pandas as pd
import pycountry
from sklearn.preprocessing import MinMaxScaler
import matplotlib.pyplot as plt

# 1. 최신 연도 데이터만 남기기
df_latest = df.sort_values('year').groupby('country').tail(1).copy()

# 2. 진짜 국가 코드(ISO alpha-3)만 남기기
real_country_codes = {c.alpha_3 for c in pycountry.countries}
df_latest = df_latest[df_latest['code'].isin(real_country_codes)].copy()

# 3. 핵심 결측치 제거
df_latest = df_latest.dropna(subset=['gdp', 'population']).copy()

# 4. 나머지 결측치는 0으로 처리
df_latest['military_usd'] = df_latest['military_usd'].fillna(0)
df_latest['hightech_exports'] = df_latest['hightech_exports'].fillna(0)
df_latest['patent_residents'] = df_latest['patent_residents'].fillna(0)

# 5. 파생 변수 생성
df_latest['gdp_per_capita'] = df_latest['gdp'] / df_latest['population']
df_latest['military_ratio'] = df_latest['military_usd'] / df_latest['gdp']
df_latest['tech_power'] = df_latest['hightech_exports'] + df_latest['patent_residents']

# 6. 정규화
scaler = MinMaxScaler()
df_latest[['gdp_norm', 'military_norm', 'tech_norm']] = scaler.fit_transform(
    df_latest[['gdp', 'military_usd', 'tech_power']]
)

# 7. 국력 점수 계산
df_latest['power_score'] = (
    df_latest['gdp_norm'] * 0.5 +
    df_latest['military_norm'] * 0.3 +
    df_latest['tech_norm'] * 0.2
)

# 8. 상위 10개국 확인
top10 = df_latest.sort_values('power_score', ascending=False).head(10)

print(top10[['country', 'code', 'power_score']])

# 9. 시각화
plt.figure(figsize=(10, 6))
plt.bar(top10['country'], top10['power_score'])
plt.xticks(rotation=45, ha='right')
plt.title('Top 10 Power Score Countries')
plt.tight_layout()
plt.show()
```
