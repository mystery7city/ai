# 🎯 숫자 맞추기 게임

터미널 대신 주피터 노트북에서 실행하는 숫자 맞추기 게임입니다.

**사용 방법:** 셀을 위에서부터 순서대로 실행하세요 (`Shift + Enter`)


```python
# ── 1. 라이브러리 & 최고 기록 함수 ──────────────────────────────
import random
import json
from pathlib import Path
from IPython.display import clear_output

RECORD_FILE = Path.home() / ".guess_the_number_record.json"

def load_record() -> dict:
    if RECORD_FILE.exists():
        with RECORD_FILE.open("r", encoding="utf-8") as f:
            return json.load(f)
    return {}

def save_record(records: dict) -> None:
    with RECORD_FILE.open("w", encoding="utf-8") as f:
        json.dump(records, f, ensure_ascii=False, indent=2)

def update_record(difficulty: str, attempts: int) -> str:
    records = load_record()
    prev = records.get(difficulty)
    if prev is None or attempts < prev:
        records[difficulty] = attempts
        save_record(records)
        return "🏆 새 최고 기록!"
    return f"(최고 기록: {prev}회)"

print("✅ 라이브러리 로드 완료")
```

```python
# ── 2. 난이도 설정 & 선택 ────────────────────────────────────────
DIFFICULTIES = {
    "1": ("쉬움",   1,  50, 10),
    "2": ("보통",   1, 100,  7),
    "3": ("어려움", 1, 200,  5),
}

def show_difficulties():
    records = load_record()
    print("=" * 40)
    print("     난이도를 선택하세요")
    print("=" * 40)
    for key, (name, low, high, max_att) in DIFFICULTIES.items():
        best = records.get(key)
        best_str = f"최고 {best}회" if best else "기록 없음"
        print(f"  [{key}] {name:6}  ({low}~{high}, 최대 {max_att}회)  {best_str}")
    print()

show_difficulties()
CHOICE = input("선택 (1/2/3): ").strip()
assert CHOICE in DIFFICULTIES, "1, 2, 3 중에서 입력해 주세요."

diff_key = CHOICE
name, low, high, max_attempts = DIFFICULTIES[diff_key]
print(f"\n✅ [{name}] 선택됨 — {low}~{high} 범위, 최대 {max_attempts}회")
```

```python
# ── 3. 게임 시작! (이 셀을 실행하면 한 판이 시작됩니다) ──────────
answer = random.randint(low, high)
attempts = 0
game_over = False

print(f"\n🎮 [{name}] 모드 시작!")
print(f"   {low}~{high} 사이의 숫자를 맞춰보세요. (최대 {max_attempts}회)\n")

while attempts < max_attempts and not game_over:
    remaining = max_attempts - attempts
    bar = "█" * remaining + "░" * attempts

    try:
        raw = input(f"[{attempts + 1}/{max_attempts}] 숫자 입력: ").strip()
        guess = int(raw)
    except ValueError:
        print("  ⚠️  숫자만 입력해 주세요.\n")
        continue

    if not (low <= guess <= high):
        print(f"  ⚠️  {low}~{high} 범위 안의 숫자를 입력해 주세요.\n")
        continue

    attempts += 1

    if guess == answer:
        result = update_record(diff_key, attempts)
        print(f"\n  🎉 정답! {attempts}회 만에 맞췄어요!  {result}")
        game_over = True
    elif guess < answer:
        print(f"  📈 너무 낮아요!   [{bar}] 남은 기회: {remaining - 1}회\n")
    else:
        print(f"  📉 너무 높아요!   [{bar}] 남은 기회: {remaining - 1}회\n")

if not game_over:
    print(f"\n  💀 아쉽! 정답은 {answer} 였어요.")
```

```python
# ── 최고 기록 보기 ───────────────────────────────────────────────
records = load_record()
print("=" * 30)
print("     🏆  최고 기록")
print("=" * 30)
if not records:
    print("  아직 기록이 없어요!")
else:
    for key, (name, low, high, _) in DIFFICULTIES.items():
        best = records.get(key, "-")
        best_str = f"{best}회" if isinstance(best, int) else best
        print(f"  {name:6}  ({low}~{high})   {best_str}")
print()
```
