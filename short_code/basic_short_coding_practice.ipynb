## 1. 리스트 컴프리헨션

리스트 컴프리헨션은 반복문을 한 줄로 줄여서 새 리스트를 만드는 방법입니다.

````python
nums = [1, 2, 3, 4, 5]
squares = [x**2 for x in nums]

print(squares)
````

위 코드는 `nums`에 있는 숫자들을 하나씩 꺼내서 제곱한 뒤, `squares`라는 새 리스트에 넣습니다.

일반 반복문으로 쓰면 아래와 같습니다.

````python
nums = [1, 2, 3, 4, 5]
squares = []

for x in nums:
    squares.append(x**2)

print(squares)
````

## 2. 조건이 있는 리스트 컴프리헨션

조건을 붙이면 원하는 값만 골라서 리스트를 만들 수 있습니다.

````python
nums = [1, 2, 3, 4, 5, 6]
evens = [x for x in nums if x % 2 == 0]

print(evens)
````

`x % 2 == 0`은 짝수인지 확인하는 조건입니다.  
따라서 위 코드는 짝수만 골라서 새 리스트를 만듭니다.

## 3. 조건식 한 줄 작성

간단한 `if-else`는 한 줄로 쓸 수 있습니다.

````python
x = 10
result = "even" if x % 2 == 0 else "odd"

print(result)
````

위 코드는 `x`가 짝수이면 `even`, 홀수이면 `odd`를 `result`에 넣습니다.

구조는 다음과 같습니다.

```python
값1 if 조건 else 값2
```

## 4. lambda 함수

`lambda`는 이름 없는 짧은 함수를 만들 때 사용합니다.

````python
add = lambda a, b: a + b

print(add(3, 5))
````

위 코드는 두 값을 더하는 함수를 한 줄로 만든 것입니다.

일반 함수로 쓰면 아래와 같습니다.

````python
def add(a, b):
    return a + b

print(add(3, 5))
````

## 5. map 사용하기

`map`은 리스트의 모든 값에 같은 함수를 적용할 때 사용합니다.

````python
nums = [1, 2, 3, 4]
doubled = list(map(lambda x: x * 2, nums))

print(doubled)
````

위 코드는 `nums`의 모든 값을 2배로 만듭니다.

다만 파이썬 입문 단계에서는 `map`보다 리스트 컴프리헨션이 더 읽기 쉬운 경우도 많습니다.

## 6. filter 사용하기

`filter`는 조건을 만족하는 값만 걸러낼 때 사용합니다.

````python
nums = [1, 2, 3, 4, 5, 6]
evens = list(filter(lambda x: x % 2 == 0, nums))

print(evens)
````

위 코드는 짝수만 남기는 코드입니다.

리스트 컴프리헨션으로 쓰면 아래와 같습니다.

````python
nums = [1, 2, 3, 4, 5, 6]
evens = [x for x in nums if x % 2 == 0]

print(evens)
````

## 7. enumerate 사용하기

`enumerate`는 반복문에서 인덱스와 값을 동시에 꺼낼 때 사용합니다.

````python
fruits = ["apple", "banana", "cherry"]

for i, fruit in enumerate(fruits):
    print(i, fruit)
````

`i`에는 순서 번호가 들어가고, `fruit`에는 리스트의 값이 들어갑니다.

## 8. zip 사용하기

`zip`은 여러 리스트를 같은 위치끼리 묶어서 사용할 때 좋습니다.

````python
names = ["Tom", "Jane", "Kim"]
scores = [90, 85, 92]

for name, score in zip(names, scores):
    print(name, score)
````

위 코드는 이름과 점수를 서로 짝지어서 출력합니다.

## 9. 딕셔너리 컴프리헨션

리스트 컴프리헨션처럼 딕셔너리도 한 줄로 만들 수 있습니다.

````python
nums = [1, 2, 3, 4]
square_dict = {x: x**2 for x in nums}

print(square_dict)
````

위 코드는 숫자를 key로, 그 숫자의 제곱을 value로 가지는 딕셔너리를 만듭니다.

## 10. sum, min, max

리스트 안의 숫자들을 빠르게 계산할 때 자주 씁니다.

````python
scores = [80, 95, 70, 100, 85]

print("합계:", sum(scores))
print("최솟값:", min(scores))
print("최댓값:", max(scores))
print("평균:", sum(scores) / len(scores))
````

`sum`은 합계, `min`은 최솟값, `max`는 최댓값을 구합니다.

## 11. 문자열 리스트 한 줄 처리

문자열 리스트에서도 쇼트 코딩을 자주 사용합니다.

````python
words = ["apple", "banana", "cherry"]
upper_words = [word.upper() for word in words]

print(upper_words)
````

위 코드는 모든 단어를 대문자로 바꿉니다.

## 12. 문자열 합치기: join

`join`은 리스트 안의 문자열을 하나로 합칠 때 사용합니다.

````python
words = ["Python", "is", "fun"]
sentence = " ".join(words)

print(sentence)
````

`" ".join(words)`는 단어 사이에 공백을 넣어서 하나의 문장으로 만듭니다.

## 13. 실전 예제: 점수 통과 여부 만들기

점수 리스트를 보고 60점 이상이면 `pass`, 아니면 `fail`로 바꿔봅니다.

````python
scores = [45, 80, 70, 55, 90]
results = ["pass" if score >= 60 else "fail" for score in scores]

print(results)
````

이 코드는 리스트 컴프리헨션과 조건식 한 줄 작성을 함께 사용한 예제입니다.

## 14. 실전 예제: 가장 많이 쓰는 패턴 정리

아래 셀은 지금까지 배운 쇼트 코딩 패턴을 한 번에 모아둔 것입니다.

````python
nums = [1, 2, 3, 4, 5]

# 제곱 리스트 만들기
squares = [x**2 for x in nums]

# 짝수만 고르기
evens = [x for x in nums if x % 2 == 0]

# 각 숫자가 짝수인지 홀수인지 표시하기
labels = ["even" if x % 2 == 0 else "odd" for x in nums]

print("squares:", squares)
print("evens:", evens)
print("labels:", labels)
````
